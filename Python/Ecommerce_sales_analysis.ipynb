```python
# Install pandas

import pandas as pd
```

```python
df = pd.read_csv("../dataset/Online Retail.csv")

df.head()
```

```python
# Initial Dataset Overview

df.shape
```

```python
df.info()
```

```python
# Check Missing Values

df.isnull().sum()
```

```python
# Removed missing values

df = df.dropna(subset=["Description"])


df.isnull().sum()
```

```python
# Check Duplicate

df.duplicated().sum()
```

```python
# Removed Duplicate

df = df.drop_duplicates()

df.duplicated().sum()
```

```python
# Converted Data Types

df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

df.info()
```

```python
# Check Negative Quantity Values

df["Quantity"].describe()

(df["Quantity"] < 0).sum()
```

```python
# Removed Negative Quantity Values

df = df[df["Quantity"] > 0]

df["Quantity"].describe()
```

```python
# Check Negative Unit Prices

df["UnitPrice"].describe()
```

```python
(df["UnitPrice"] <= 0).sum()
```

```python
#  Removed Negative Unit Prices

df = df[df["UnitPrice"] > 0]

df["UnitPrice"].describe()
```

```python
# 9. Create Revenue Column

df["Revenue"] = df["Quantity"] * df["UnitPrice"]

df.head()
```

```python
# 10. Final Dataset Check  

df.shape
```

```python
#   Export to SQL

from sqlalchemy import create_engine

engine = create_engine(
    "postgresql://postgres:vasu170402@localhost:5432/ecommerce_sales_analysis"
)

df.to_sql(
    "ecommerce_sales_analysis",
    engine,
    if_exists="replace",
    index=False
)
```
